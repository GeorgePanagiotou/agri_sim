```python
from pathlib import Path
import pandas as pd
from agrisim.config import REPO
data_root = REPO / 'data/processed'
print('Project:', REPO)
print('Processed files:')
for path in sorted(data_root.rglob('*')):
    if path.is_file() and path.suffix in {'.csv','.parquet','.nc','.tif'}:
        print(path.relative_to(REPO), f'{path.stat().st_size / 1024**2:.2f} MiB')
```

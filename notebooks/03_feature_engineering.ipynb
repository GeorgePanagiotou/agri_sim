```python
from pathlib import Path
import pandas as pd
from agrisim.features.weather import add_rainfall_features

source = Path("data/processed/power_daily_20230101_20251231.parquet")

df = pd.read_parquet(source)

df = add_rainfall_features(df)

output = Path(
    "data/processed/features/power_weather_features_2023_2025.parquet"
)
output.parent.mkdir(parents=True, exist_ok=True)

df.to_parquet(output)

print(df[[
    "rainfall_prev_7d_mm",
    "rainfall_prev_14d_mm",
    "rainfall_prev_30d_mm"
]].tail())
```

# 02 — Sowing / risk / rotation decision engine
Regional NASA environmental data + transparent decision rules. This is a historical evidence explorer, not a calibrated farm-sowing forecast.

```python
from pathlib import Path
import pandas as pd
ROOT=Path.cwd()
assert (ROOT/'sources.json').exists(), 'Run Jupyter in Agrisimfinal project root'
from agrisim.features.build_features import run as build_features
from agrisim.decision_engine.sowing_rules import recommend_sowing
from agrisim.decision_engine.risk_assessment import risk_assessment
from agrisim.decision_engine.ranking import rank_rotations
```

```python
power=list((ROOT/'data/processed').glob('power_daily_*.parquet'))
if power:
    candidates=build_features(ROOT/'data/reference/fields.csv')
    display(recommend_sowing(candidates)[['field_id','candidate_week_start','sowing_screen','sowing_explanation']].tail(10))
    display(risk_assessment(candidates)[['field_id','candidate_week_start','risk_level']].tail(10))
else:
    print('NASA POWER data not yet downloaded. See README.md for approved download commands.')
```

```python
rot=pd.read_csv(ROOT/'data/reference/rotations_template.csv')
result=rank_rotations(rot)
display(result[['strategy_id','rotation','evidence_status','priority_rank']])
print('Blank local scores correctly remain unranked until a farmer/agronomist supplies them.')
```
